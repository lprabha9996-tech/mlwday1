Feature Scaling -->Min Max Normalization
x(Scaled)=x(i) - x(min)/x(min)

```python
pip install scikit-learn
```

```python
#Import the libraries
import numpy as np
import pandas as pd
import sklearn
```

```python
data = pd.DataFrame({'Price_USA':[80,50,140,25],
                     'Price_INR':[5000,7000,12000,3000]})
data
```

```python
data['Price_USA_minmax'] = (data['Price_USA'] - data['Price_USA'].min()) /(data['Price_USA'].max() -data['Price_USA'].min())
data
```

```python
data['Price_INR_minmax'] = (data['Price_INR'] - data['Price_INR'].min()) /(data['Price_INR'].max() -data['Price_INR'].min())
data
```

```python
import sklearn
```

```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler
```

```python
min_max= MinMaxScaler()
```

```python
data['Price_USA_minsk']=min_max.fit_transform(data[['Price_USA']])
data
```

```python
data['Price_INR_minsk']=min_max.fit_transform(data[['Price_INR']])
data
```

```python
data['Price_USA_std'] = (data['Price_USA'] - data['Price_USA'].mean())/(data['Price_USA'].std())
data
```

```python
data['Price_INR_std'] = (data['Price_INR'] - data['Price_INR'].mean())/(data['Price_INR'].std())
data
```

```python
scaler = StandardScaler()
```

```python
data['Price_USA_Scaler'] = scaler.fit_transform(data[['Price_USA']])
data
```

```python
data['Price_INR_Scaler'] = scaler.fit_transform(data[['Price_INR']])
data
```

```python
from sklearn.preprocessing import RobustScaler
```

```python
robust = RobustScaler()
```

```python
data['Price_USA_robust'] = robust.fit_transform(data[['Price_USA']])
data['Price_USA_robust']
```

```python
data['Price_INR_robust'] = robust.fit_transform(data[['Price_INR']])
data['Price_INR_robust']
```

##Regression on Case Study related to predicting Miles Per Gallon (mog)
#Sceniario as follows :

A Fleet manager wanted to calculate (predict) the mpg efficiency of his vehicles based on (cylinders, weight, displacement, acceleration, horsepower)...

```python
pip install ucimlrepo
```

```python
##Step:1 : Importing the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from ucimlrepo import fetch_ucirepo
from sklearn.linear_model import LinearRegression
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import train_test_split,GridSearchCV
from sklearn.preprocessing import OneHotEncoder,PolynomialFeatures
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
##Step2: Loading (autompg dataset) and Cleaning the Data
data = fetch_ucirepo(id=9)
data
x =data.data.features.copy()
x
```

```python
y=data.data.targets['mpg']
```

```python
x['origin'].unique();
x['horsepower'].unique()
```

```python

x.info()
numerical_features=['cylinders','displacement','horsepower','weight','acceleration','model_year']
categorical_features = ['origin']
x=x[numerical_features+categorical_features]
#x.info()
#x.isnull().sum()
x.shape
```

```python
#first we will trnasform the horsepower and origin columns
x['horsepower'] = pd.to_numeric(x['horsepower'],errors='coerce')
#X['horsepower'].info()
x['origin'] = x['origin'].astype(str)
```

```python
#step3: Quick exploratory data analysis (seaborn)
df = x.copy()
df['mpg']=y
```

```python
#DF
#we are gonna use sunplots() concept to divide the plot into multiple plotes
fig,axes = plt.subplots(1,3,figsize=(16,4))
#plot graphs as per axes
sns.histplot(df['mpg'],kde=True,ax=axes[0],color='purple')
#in similar way we display other graphs in rmg plots
axes[0].set_title("MPG Distribution")
sns.scatterplot(x=df['weight'],y=df['mpg'],ax=axes[1],color='blue')
axes[1].set_title("Weight vs MPG Distribution")
sns.scatterplot(x=df['horsepower'],y=df['mpg'],ax=axes[2],color='green')
axes[2].set_title("HorsePower vs MPG Distribution")
```

```python
#Correlation between our variables
plt.figure(figsize=(16,4))
sns.heatmap(df.select_dtypes(include=np.number).corr(),
             annot=True,
            cmap='coolwarm',
            center=0
            )
plt.title("Correlation matrix")
plt.tight_layout()
plt.show()
```

```python
#Step 4: Divivding the data into train test split
#print(x)
#print(y)
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
print('\n Train test Split')
print("Training Data",len(x_train))
print("Testing Data",len(x_test))
```

```python
#step 5 :creating an evalutations function for baseline model
def evaluate(model_name,actual,predicted):
    """Returns MAE,MSE,RMSE,R2-Socre"""
    mse=mean_squared_error(actual,predicted)
    mae=mean_absolute_error(actual , predicted)
    return {'Model_Name':model_name,
    'MAE':mae,
    'MSE':mse,
    'RMSE':np.sqrt(mse),
    'R2-Score':r2_score(actual,predicted)}
results=[]
```

```python
#Step 6 : Baseline model -->It predicts the training set for every values
baseline_predictions =np.full(len(y_test),y_train.mean())
#baseline_predictions
#in above case we are keeping a baseline prediction
results.append(evaluate("Mean baseline",y_test,baseline_predictions))
results
```

```python
#step7:We apply Linear Regression
simple_model =LinearRegression()
#we fit the data model
simple_model.fit(x_train[['weight']],y_train)
simple_model_predictions =simple_model.predict(x_test[['weight']])
#print(simple_model_predictions)
results.append(evaluate("Simple Linear Regression:Weight",
                        y_test,
                        simple_model_predictions))
print(len(results))
```

```python
results
```

```python
#we will check the intercept value
print(f'Simple Linear Regression Equation')
print(f"Predicted MPG = {simple_model.intercept_}+({simple_model.coef_[0]}*weight")
```
