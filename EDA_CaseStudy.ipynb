```python
print(data['Age'].mean())
```

```python
#We can fill the Age column missing values with mean
data['Age'] = data['Age'].fillna(data['Age'].mean(),inplace=True);
data.isnull().sum()
```

```python
data['Embarked'] = data['Embarked'].fillna('S',inplace=True)
```

```python
data.drop(columns=['Cabin'],inplace=True)
```

```python
data.isnull().sum()
```

```python
pclass=pd.get_dummies(data['Pclass'],
                      drop_first=True,
                      dtype=int)
pclass
```

```python
gender=pd.get_dummies(data['Sex'],dtype=int)
gender
```

```python
embarked=pd.get_dummies(data['Embarked'],dtype=int)
embarked
```

```python
final_data=pd.concat([data,pclass,gender,embarked],
                     axis=1)
final_data
```

```python
final_data.drop(columns=['PassengerId','Name',
                         'Pclass','Sex','Ticket',
                         'Embarked'],inplace=True)
final_data.head()
```
